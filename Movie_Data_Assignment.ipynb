### **Question 36**

### **Question 40**

(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
import matplotlib.pyplot as plt
import numpy as np

top_5 = df.nlargest(5, 'IMDB_Rating').copy()

top_5['Rating_100_Scale'] = top_5['IMDB_Rating'] * 10

x = np.arange(len(top_5))
width = 0.35

plt.figure(figsize=(12, 6))

plt.bar(x - width/2, top_5['Rating_100_Scale'],
        width, label='IMDb Rating (100 Scale)')

plt.bar(x + width/2, top_5['Meta_score'],
        width, label='Meta Score')

plt.xticks(x, top_5['Series_Title'], rotation=45, ha='right')
plt.title('Top 5 Movies: IMDb Rating vs Meta Score')
plt.xlabel('Movie Name')
plt.ylabel('Score (Out of 100)')
plt.legend()
plt.tight_layout()
plt.show()
```
